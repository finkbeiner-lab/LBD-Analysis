## Import Libraries

```python
import pandas as pd
import numpy as np
import os
import glob
import shutil
import pdb
```

## Load Input

```python
val_dir = "/gladstone/finkbeiner/steve/work/data/npsad_data/monika/LBD/Train_val_LB/val/"
val_dirs = glob.glob(os.path.join(val_dir, "*.svs"))
ground_truth_dataset=pd.DataFrame()
for v in val_dirs:
    l = glob.glob(os.path.join(v, "images", "*"))
    if len(ground_truth_dataset)==0:
        ground_truth_dataset = pd.DataFrame({"image_path":l})
    else:
        ground_truth_dataset = pd.concat([ground_truth_dataset, pd.DataFrame({"image_path":l})],ignore_index=True)
    
```

```python
ground_truth_dataset["image_name"]=ground_truth_dataset["image_path"].apply(lambda l:l.split("/")[-1])
ground_truth_dataset["image_name"] =  ground_truth_dataset["image_name"].apply(lambda l: l.replace(".png",""))
ground_truth_dataset["image_name"] =  ground_truth_dataset["image_name"].apply(lambda l: l.replace(".svs",""))
```

```python
ground_truth_dataset["wsi_id"]=ground_truth_dataset["image_path"].apply(lambda l:l.split("/")[-1].split("_CG_")[0])
```

```python
ground_truth_dataset["ground_truth"]=True
```

```python
result_data_path = "/gladstone/finkbeiner/steve/work/data/npsad_data/monika/LBD/Reports/figures/super-disco-93/"
result_data_path = "/gladstone/finkbeiner/steve/work/data/npsad_data/monika/LBD/Reports/figures/royal-butterfly-145_mrcnn_model_10/"
result_folders = glob.glob(os.path.join(result_data_path, "*.svs"))
print(result_folders)
#destination_path = "/home/vivek/Projects/amyb-plaque-detection/reports/"
#csv_dir = os.path.join(os.path.dirname(destination_path), "csv_data")
#if not os.path.exists(csv_dir):
#    os.makedirs(csv_dir)
final_csv_merged = pd.DataFrame()
for result_folder in result_folders:
    csv_file = glob.glob(os.path.join(result_folder, "*.csv"))
    print(csv_file)
    print(len(csv_file))
    if (len(csv_file) == 0) or (os.path.getsize(csv_file[0])<2):
        print("Empty", result_folder)
        continue
    
    if len(csv_file)>1:
        print(">1 csv files")
        break
    
    if len(csv_file)==1:
        csv_file = pd.read_csv(csv_file[0])
        print(len(csv_file))
        
    
    
    if (len(final_csv_merged)==0):
        final_csv_merged=csv_file
    else:
        final_csv_merged=pd.concat([final_csv_merged,csv_file])
    
```
