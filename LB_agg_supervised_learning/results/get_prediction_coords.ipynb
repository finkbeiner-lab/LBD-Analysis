```python
folders= os.listdir(output_path)
folders.remove(".DS_Store")
```

```python
csv_file_list = []
for folder in folders:
    csv_file = glob.glob(os.path.join(output_path,folder,"*.csv"))[0]
    csv_file_list.append(csv_file)
```

```python
coords_path = "/gladstone/finkbeiner/steve/work/data/npsad_data/monika/LBD/Testing_results/wgm_model_crops_pred_LBs/coords"
for csv_file in csv_file_list:
    csv_file_pd = pd.read_csv(csv_file)
    csv_file_pd["x"] = csv_file_pd["image_name"].apply(lambda l: int(float(l.split("_x_")[1].split("_")[0])))
    csv_file_pd["y"] = csv_file_pd["image_name"].apply(lambda l: int(float(l.split("_y_")[1].split("_")[0].split(".")[0])))
    csv_file_pd["centroid_x"] = csv_file_pd["centroid"].apply(lambda l: int(float(l.split(",")[0].split("(")[1])))
    csv_file_pd["centroid_y"] = csv_file_pd["centroid"].apply(lambda l: int(float(l.split(",")[1].split(")")[0])))
    csv_file_pd["x_centroid_x"] = (csv_file_pd["x"] + csv_file_pd["centroid_x"])
    csv_file_pd["y_centroid_y"] = (csv_file_pd["y"] + csv_file_pd["centroid_y"])
    csv_file_pd[["x_centroid_x","y_centroid_y"]].to_csv(os.path.join(coords_path, csv_file.split("/")[-1]),index=False)
```
